```python
import pyspark

import requests
import pandas as pd
from sodapy import Socrata
from pymongo import MongoClient
import json
```

```python
with open("secrets_socrata.json", "r") as f:
    json_secrets = json.load(f)
```

```python
def get_data_iowa_liquors(secrets, date_query="2025-10-30", row_limit=99999):
    """
    Connect to Socrata API and return a json
    """
    client = Socrata("data.iowa.gov",
                 json_secrets["secrets"]["APP_TOKEN"],
                 username=json_secrets["login"]["username"],
                 password=json_secrets["login"]["password"])
    results = client.get("m3tr-qhgy", date=date_query, limit=row_limit)
    return results
```

```python
get_data_iowa_liquors(json_secrets)
```
